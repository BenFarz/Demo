```python
# Import necessary libraries
!pip install ultralytics
from google.colab import drive
import os
from skimage import io
import numpy as np
from matplotlib import pyplot as plt
from ultralytics import YOLO

from PIL import Image
import yaml
import tensorflow as tf

# Mount Google Drive
drive.mount('/content/drive')
```

```python
# Set paths for training, validation, and configuration
train_path = '/content/drive/MyDrive/Tr120-Val30/train'
valid_path = '/content/drive/MyDrive/Tr120-Val30/valid'
yaml_path = '/content/drive/MyDrive/Tr120-Val30/data.yaml'
```

```python
# Load the configuration from the YAML file and print it
with open(yaml_path, 'r') as stream:
    data = yaml.safe_load(stream)
    print(data)
```

```python
# Define project directory for results
project = "/content/drive/MyDrive/Tr120-Val30/results"
# Define subdirectory name for this specific training
name = "200_epochs"  # This will create a new directory like '200_epochs-', '200_epochs-2', etc.
```

```python
# Load the pretrained model and initialize for transfer learning
model = YOLO('yolov8n-seg.pt')  # Transfer the weights from a pretrained model
```

```python
# Train the model
results = model.train(data=yaml_path,
                      project=project,
                      name=name,
                      epochs=200,
                      patience=0,  # Disable early stopping
                      batch=4,
                      imgsz=640)  # Image size in pixels (assumed to be square images)

# Copy logs to Google Drive for persistent storage
!cp -r runs/train /content/drive/MyDrive/Tr120-Val30/runs
```
